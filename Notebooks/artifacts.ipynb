```python
from sklearn.model_selection import train_test_split
import os, shutil

# Example: random split of dataset
source_dir = "D:/Project/Waste management/data/all images"
train_dir = "D:/Project/Waste management/data/train"
test_dir = "D:/Project/Waste management/data/test"

# For each class, split files 80/20
for cls in os.listdir(source_dir):
    cls_path = os.path.join(source_dir, cls)
    images = os.listdir(cls_path)
    train_imgs, test_imgs = train_test_split(images, test_size=0.2, random_state=42)
    os.makedirs(os.path.join(train_dir, cls), exist_ok=True)
    os.makedirs(os.path.join(test_dir, cls), exist_ok=True)
    for img in train_imgs:
        shutil.copy(os.path.join(cls_path, img), os.path.join(train_dir, cls, img))
    for img in test_imgs:
        shutil.copy(os.path.join(cls_path, img), os.path.join(test_dir, cls, img))
```
